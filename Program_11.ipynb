File Allocation Methods — <b>Contiguous</b>, <b>Linked</b> and <b>Indexed Allocation</b>.
</p>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>AIM:</b> To simulate Contiguous, Linked and Indexed file allocation methods used by the operating system to allocate disk blocks to files, and to display the block allocation table for each.
</p>



<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li> Contiguous allocation stores each file as a set of contiguous blocks on the disk. It gives excellent sequential and direct access performance but suffers from external fragmentation and requires knowing the file size in advance.</li>
<li>Linked allocation stores each file as a linked list of disk blocks scattered anywhere on disk; each block contains a pointer to the next block. It eliminates external fragmentation but only supports efficient sequential access, and pointers consume space and are vulnerable to corruption.</li>
<li>Indexed allocation brings all block pointers of a file together into one index block. This supports efficient direct access and avoids external fragmentation, at the cost of extra space for the index block.</li>
</ul>
</div>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>Program(PYTHON) and OUTPUT:</b>
</p>


```python
MAXBLOCKS = 30
disk = [0] * MAXBLOCKS      # 0 = free, 1 = allocated


def reset_disk():
    for i in range(MAXBLOCKS):
        disk[i] = 0


def contiguous_allocation():
    reset_disk()
    print("\n--- Contiguous Allocation ---")
    nf = int(input("Enter number of files: "))
    for f in range(1, nf + 1):
        start, length = map(int, input(
            f"File {f} - starting block, length: ").split())
        ok = all(0 <= i < MAXBLOCKS and disk[i] == 0
                 for i in range(start, start + length))
        if not ok:
            print("Allocation failed: blocks not free/out of range.")
            continue
        for i in range(start, start + length):
            disk[i] = 1
        print(f"File {f} allocated blocks {start} to {start + length - 1}")


def linked_allocation():
    reset_disk()
    print("\n--- Linked Allocation ---")
    nf = int(input("Enter number of files: "))
    for f in range(1, nf + 1):
        length = int(input(f"File {f} - number of blocks: "))
        blocks = []
        for i in range(MAXBLOCKS):
            if len(blocks) == length:
                break
            if disk[i] == 0:
                disk[i] = 1
                blocks.append(i)
        print(f"File {f} allocated blocks: " +
              " -> ".join(map(str, blocks)) + " -> (NULL)")
        if len(blocks) < length:
            print("Warning: disk full, could not allocate all blocks.")


def indexed_allocation():
    reset_disk()
    print("\n--- Indexed Allocation ---")
    nf = int(input("Enter number of files: "))
    for f in range(1, nf + 1):
        length = int(input(f"File {f} - number of data blocks needed: "))
        index_block = -1
        for i in range(MAXBLOCKS):
            if disk[i] == 0:
                index_block = i
                disk[i] = 1
                break
        blocks = []
        for i in range(MAXBLOCKS):
            if len(blocks) == length:
                break
            if disk[i] == 0:
                disk[i] = 1
                blocks.append(i)
        print(f"File {f} - Index block: {index_block}, Data blocks: "
              + " ".join(map(str, blocks)))


def main():
    print("Menu:\n1. Contiguous Allocation\n2. Linked Allocation")
    print("3. Indexed Allocation")
    choice = int(input("Enter choice: "))
    if choice == 1:
        contiguous_allocation()
    elif choice == 2:
        linked_allocation()
    elif choice == 3:
        indexed_allocation()
    else:
        print("Invalid choice")


if __name__ == "__main__":
    main()
```
