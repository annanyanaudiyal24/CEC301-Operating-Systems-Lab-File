<p style="font-family: Times New Roman;
font-size: 18px;">
<b>AIM:</b> To implement the Banker's Algorithm to check whether a system is in a safe state, and to simulate resource allocation requests, thereby avoiding deadlock.
</p>



<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li> The Banker's Algorithm is a deadlock-avoidance algorithm that tests whether granting a resource request leaves the system in a safe state before actually allocating resources.</li>
<li>It uses four data structures: Available (resources currently free), Max (maximum demand of each process), Allocation (resources currently allocated to each process), and Need = Max − Allocation.</li>
<li>The Safety Algorithm finds a sequence in which all processes can finish without deadlock by repeatedly finding a process whose Need can be satisfied by Available, then simulating its completion (adding its allocation back to Available).</li>
<li>The Resource-Request Algorithm grants a request only if the resulting state (obtained by tentatively allocating) is still safe; otherwise the requesting process must wait.</li>
</ul>
</div>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>Program(PYTHON) and OUTPUT:</b>
</p>


```python
P = 5   # number of processes
R = 3   # number of resource types

available = [0] * R
maximum = [[0] * R for _ in range(P)]
allocation = [[0] * R for _ in range(P)]
need = [[0] * R for _ in range(P)]


def calculate_need():
    for i in range(P):
        for j in range(R):
            need[i][j] = maximum[i][j] - allocation[i][j]


def is_safe():
    """Safety algorithm. Returns (True, safe_sequence) or (False, [])."""
    work = available[:]
    finish = [False] * P
    safe_seq = []
    while len(safe_seq) < P:
        found = False
        for i in range(P):
            if not finish[i] and all(need[i][j] <= work[j] for j in range(R)):
                for k in range(R):
                    work[k] += allocation[i][k]
                safe_seq.append(i)
                finish[i] = True
                found = True
        if not found:
            return False, []            # no process could proceed: unsafe
    return True, safe_seq


def request_resources(pid, request):
    for j in range(R):
        if request[j] > need[pid][j]:
            print("Error: process has exceeded its maximum claim.")
            return False
        if request[j] > available[j]:
            print("Request denied: not enough resources available.")
            return False

    for j in range(R):                  # tentatively allocate
        available[j] -= request[j]
        allocation[pid][j] += request[j]
        need[pid][j] -= request[j]

    safe, _ = is_safe()
    if safe:
        print("Request can be granted safely.")
        return True

    for j in range(R):                  # rollback
        available[j] += request[j]
        allocation[pid][j] -= request[j]
        need[pid][j] += request[j]
    print("Request denied: would lead to an unsafe state.")
    return False


def main():
    global available
    available = list(map(int, input(
        f"Enter available resources of each type ({R} types): ").split()))

    print(f"Enter Maximum matrix ({P} processes x {R} resources):")
    for i in range(P):
        maximum[i] = list(map(int, input().split()))

    print("Enter Allocation matrix:")
    for i in range(P):
        allocation[i] = list(map(int, input().split()))

    calculate_need()
    safe, seq = is_safe()
    if safe:
        print("System is in a SAFE state.")
        print("Safe sequence:", " ".join(f"P{i}" for i in seq))
    else:
        print("System is in an UNSAFE state.")

    pid = int(input("\nEnter process number requesting resources: "))
    request = list(map(int, input(f"Enter request vector ({R} values): ").split()))
    request_resources(pid, request)


if __name__ == "__main__":
    main()
```
