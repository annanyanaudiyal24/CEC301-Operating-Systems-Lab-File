<b>AIM:</b> To implement and compare non-preemptive First Come First Served (FCFS), Shortest Job First (SJF), Priority scheduling, and preemptive Round Robin CPU scheduling algorithms, and compute average waiting time and turnaround time.
</p>



<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li>CPU scheduling decides which of the ready processes is allocated the CPU. FCFS executes processes strictly in order of arrival — simple but can cause the convoy effect.</li>
<li>SJF (non-preemptive) selects the process with the smallest burst time among the available processes; it minimizes average waiting time but can cause starvation of longer jobs.</li>
<li>Priority scheduling allocates the CPU to the process with the highest priority (here, lower number = higher priority); it can also cause starvation, solved using aging.</li>
<li>Round Robin allocates the CPU to each ready process for a fixed time quantum in cyclic order — it is preemptive and provides fair, responsive scheduling suitable for time-sharing systems.</li>
<li>Waiting Time = Turnaround Time − Burst Time,  Turnaround Time = Completion Time − Arrival Time.</li>
</ul>
</div>

<b>First Come First Serve (FCFS):</b>
</p>


```python
def print_result(p):
    n = len(p)
    print("\nPID\tAT\tBT\tCT\tTAT\tWT")
    for x in p:
        print(f"{x['pid']}\t{x['at']}\t{x['bt']}\t{x['ct']}\t{x['tat']}\t{x['wt']}")
    print(f"Average Waiting Time    = {sum(x['wt'] for x in p) / n:.2f}")
    print(f"Average Turnaround Time = {sum(x['tat'] for x in p) / n:.2f}")


def finish(x, time):
    x["ct"] = time
    x["tat"] = x["ct"] - x["at"]
    x["wt"] = x["tat"] - x["bt"]


def fcfs(p):
    time = 0
    for x in p:
        if time < x["at"]:
            time = x["at"]
        time += x["bt"]
        finish(x, time)
    print_result(p)


def main():
    n = int(input("Enter number of processes: "))
    p = []
    for i in range(n):
        at, bt = map(int, input(f"Process {i + 1} - Arrival Time, Burst Time: ").split())
        p.append({"pid": i + 1, "at": at, "bt": bt, "ct": 0, "tat": 0, "wt": 0})
    fcfs(p)


if __name__ == "__main__":
    main()
```

<b>Shortest Job First (SJF):</b>
</p>


```python
def print_result(p):
    n = len(p)
    print("\nPID\tAT\tBT\tCT\tTAT\tWT")
    for x in p:
        print(f"{x['pid']}\t{x['at']}\t{x['bt']}\t{x['ct']}\t{x['tat']}\t{x['wt']}")
    print(f"Average Waiting Time    = {sum(x['wt'] for x in p) / n:.2f}")
    print(f"Average Turnaround Time = {sum(x['tat'] for x in p) / n:.2f}")


def finish(x, time):
    x["ct"] = time
    x["tat"] = x["ct"] - x["at"]
    x["wt"] = x["tat"] - x["bt"]


def sjf(p):
    n = len(p)
    done = [False] * n
    completed = time = 0
    while completed < n:
        idx, min_bt = -1, float("inf")
        for i in range(n):
            if not done[i] and p[i]["at"] <= time and p[i]["bt"] < min_bt:
                min_bt, idx = p[i]["bt"], i
        if idx == -1:                   # no process has arrived yet: CPU idle
            time += 1
            continue
        time += p[idx]["bt"]
        finish(p[idx], time)
        done[idx] = True
        completed += 1
    print_result(p)


def main():
    n = int(input("Enter number of processes: "))
    p = []
    for i in range(n):
        at, bt = map(int, input(f"Process {i + 1} - Arrival Time, Burst Time: ").split())
        p.append({"pid": i + 1, "at": at, "bt": bt, "ct": 0, "tat": 0, "wt": 0})
    sjf(p)


if __name__ == "__main__":
    main()
```

```python
def print_result(p):
    n = len(p)
    print("\nPID\tAT\tBT\tCT\tTAT\tWT")
    for x in p:
        print(f"{x['pid']}\t{x['at']}\t{x['bt']}\t{x['ct']}\t{x['tat']}\t{x['wt']}")
    print(f"Average Waiting Time    = {sum(x['wt'] for x in p) / n:.2f}")
    print(f"Average Turnaround Time = {sum(x['tat'] for x in p) / n:.2f}")


def finish(x, time):
    x["ct"] = time
    x["tat"] = x["ct"] - x["at"]
    x["wt"] = x["tat"] - x["bt"]


def priority_scheduling(p):
    n = len(p)
    done = [False] * n
    completed = time = 0
    while completed < n:
        idx, best = -1, float("inf")
        for i in range(n):
            if not done[i] and p[i]["at"] <= time and p[i]["pr"] < best:
                best, idx = p[i]["pr"], i
        if idx == -1:
            time += 1
            continue
        time += p[idx]["bt"]
        finish(p[idx], time)
        done[idx] = True
        completed += 1
    print_result(p)


def main():
    n = int(input("Enter number of processes: "))
    p = []
    for i in range(n):
        at, bt, pr = map(int, input(
            f"Process {i + 1} - Arrival Time, Burst Time, Priority: ").split())
        p.append({"pid": i + 1, "at": at, "bt": bt, "pr": pr, "ct": 0, "tat": 0, "wt": 0})
    priority_scheduling(p)


if __name__ == "__main__":
    main()
```

<b>Round Robin:</b>
</p>


```python
from collections import deque

def print_result(p):
    n = len(p)
    print("\nPID\tAT\tBT\tCT\tTAT\tWT")
    for x in p:
        print(f"{x['pid']}\t{x['at']}\t{x['bt']}\t{x['ct']}\t{x['tat']}\t{x['wt']}")
    print(f"Average Waiting Time    = {sum(x['wt'] for x in p) / n:.2f}")
    print(f"Average Turnaround Time = {sum(x['tat'] for x in p) / n:.2f}")


def finish(x, time):
    x["ct"] = time
    x["tat"] = x["ct"] - x["at"]
    x["wt"] = x["tat"] - x["bt"]


def round_robin(p, quantum):
    n = len(p)
    time = completed = 0
    in_queue = [False] * n
    queue = deque()
    for i in range(n):                  # processes present at time 0
        if p[i]["at"] <= time:
            queue.append(i)
            in_queue[i] = True

    while completed < n:
        if not queue:                   # CPU idle
            time += 1
            for i in range(n):
                if not in_queue[i] and p[i]["rem"] > 0 and p[i]["at"] <= time:
                    queue.append(i)
                    in_queue[i] = True
            continue
        idx = queue.popleft()
        sl = min(p[idx]["rem"], quantum)
        time += sl
        p[idx]["rem"] -= sl
        for i in range(n):              # newly arrived processes join before idx is re-queued
            if not in_queue[i] and p[i]["rem"] > 0 and p[i]["at"] <= time and i != idx:
                queue.append(i)
                in_queue[i] = True
        if p[idx]["rem"] > 0:
            queue.append(idx)
        else:
            finish(p[idx], time)
            completed += 1
            in_queue[idx] = False
    print_result(p)


def main():
    n = int(input("Enter number of processes: "))
    p = []
    for i in range(n):
        at, bt = map(int, input(f"Process {i + 1} - Arrival Time, Burst Time: ").split())
        p.append({"pid": i + 1, "at": at, "bt": bt, "ct": 0, "tat": 0, "wt": 0, "rem": bt})
    quantum = int(input("Enter time quantum: "))
    round_robin(p, quantum)


if __name__ == "__main__":
    main()
```
