<p><b>Theory:</b></p>
<ul>
<li> A process is a program in execution. Every process in UNIX/Linux is identified by a unique Process ID (PID). The operating system provides system calls to create, control and terminate processes.</li>
<li>fork() creates a new process (child) which is an almost exact copy of the calling (parent) process. It returns 0 in the child, the child's PID in the parent, and -1 on failure.</li>
<li>The exec() family of functions (execl, execv, execlp, execvp, execve) replaces the current process image with a new program. The PID does not change; only the code, data and stack are replaced.</li>
<li>wait()/waitpid() suspends execution of the calling process until one of its children terminates, and allows the parent to collect the child's exit status, preventing zombie processes.</li>
<li>getpid() returns the process ID of the calling process; getppid() returns the parent's process ID.</li>
</ul>

```python
import os
import sys
import warnings

warnings.filterwarnings("ignore",message="This process.*use of fork\\(\\) may lead to deadlocks.*",
    category=DeprecationWarning)
print(f"Parent process: PID = {os.getpid()}")
sys.stdout.flush()          # flush before fork so output isn't duplicated

try:
    pid = os.fork()
except OSError as e:
    print("fork failed:", e)
    sys.exit(1)

if pid == 0:
    # Child process
    print(f"Child process: PID = {os.getpid()}, Parent PID = {os.getppid()}")
    print("Child is about to exec 'ls -l'")
    sys.stdout.flush()
    try:
        os.execvp("ls", ["ls", "-l"])      # replaces the child's process image
    except OSError as e:                   # only reached on error
        print("exec failed:", e)
        os._exit(1)
else:
    # Parent process
    print(f"Parent waiting for child (PID = {pid}) to finish...")
    sys.stdout.flush()
    _, status = os.wait()
    if os.WIFEXITED(status):
        print(f"Child exited with status {os.WEXITSTATUS(status)}")
    print("Parent process resumes and exits.")
```
