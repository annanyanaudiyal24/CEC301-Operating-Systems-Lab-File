<b>AIM:</b> To implement FCFS, SSTF, SCAN and C-SCAN disk scheduling algorithms and compute the total head movement for a given set of disk track requests.
</p>



<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li> Disk scheduling determines the order in which pending I/O requests (cylinder/track numbers) are serviced by the disk arm, to minimize total seek time (head movement).</li>
<li>FCFS services requests strictly in the order they arrive — fair but can result in a large total head movement.</li>
<li>SSTF (Shortest Seek Time First) always services the request closest to the current head position — reduces seek time but can cause starvation of far-away requests.</li>
<li>SCAN (elevator algorithm) moves the disk arm in one direction, servicing all requests along the way, until it reaches the end, then reverses direction.</li>
<li>C-SCAN (Circular SCAN) moves the arm in one direction servicing requests, then jumps back to the beginning without servicing on the return trip, giving more uniform wait times.</li>
</ul>
</div>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>Program(PYTHON) and OUTPUT:</b>
</p>


```python
def fcfs(req, head):
    total, cur = 0, head
    for r in req:
        total += abs(cur - r)
        cur = r
    print(f"FCFS   total head movement = {total}")


def sstf(req, head):
    visited = [False] * len(req)
    total, cur = 0, head
    for _ in range(len(req)):
        idx, best = -1, float("inf")
        for i, r in enumerate(req):
            if not visited[i] and abs(cur - r) < best:
                best, idx = abs(cur - r), i
        visited[idx] = True
        total += best
        cur = req[idx]
    print(f"SSTF   total head movement = {total}")


def scan(req, head, disk_size):
    arr = sorted(req + [head])
    idx = arr.index(head)
    total, cur = 0, head
    for i in range(idx + 1, len(arr)):          # towards the higher end first
        total += abs(cur - arr[i])
        cur = arr[i]
    total += abs(cur - (disk_size - 1))         # go to the end
    cur = disk_size - 1
    for i in range(idx - 1, -1, -1):            # then reverse
        total += abs(cur - arr[i])
        cur = arr[i]
    print(f"SCAN   total head movement = {total}")


def cscan(req, head, disk_size):
    arr = sorted(req + [head])
    idx = arr.index(head)
    total, cur = 0, head
    for i in range(idx + 1, len(arr)):
        total += abs(cur - arr[i])
        cur = arr[i]
    total += abs(cur - (disk_size - 1))         # go to the end
    total += disk_size - 1                      # jump to cylinder 0
    cur = 0
    for i in range(idx):
        total += abs(cur - arr[i])
        cur = arr[i]
    print(f"C-SCAN total head movement = {total}")


def main():
    n = int(input("Enter number of requests: "))
    req = list(map(int, input("Enter the requests: ").split()))[:n]
    head = int(input("Enter initial head position: "))
    disk_size = int(input("Enter disk size (max cylinder + 1): "))

    fcfs(req, head)
    sstf(req, head)
    scan(req, head, disk_size)
    cscan(req, head, disk_size)


if __name__ == "__main__":
    main()
```
