Page Replacement Algorithms — <b>FIFO</b>, <b>LRU</b> and <b>Optimal</b>.
</p>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>AIM:</b> To implement and compare the FIFO, Least Recently Used (LRU) and Optimal page replacement algorithms used in demand paging, and to count the number of page faults produced by each for a given reference string.
</p>



<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li> In demand paging, when a referenced page is not in memory (a page fault) and all frames are full, the OS must select a victim page to replace, based on a page-replacement algorithm.</li>
<li>FIFO replaces the page that has been in memory the longest, regardless of usage — simple but can suffer from Belady's Anomaly (more frames causing more faults).</li>
<li>LRU replaces the page that has not been used for the longest period of time — approximates optimal behaviour based on the principle of locality of reference.</li>
<li>The Optimal (MIN) algorithm replaces the page that will not be used for the longest time in the future — it gives the theoretical minimum number of page faults but requires future knowledge and is used only as a benchmark.</li>
</ul>
</div>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>Program(PYTHON) and OUTPUT:</b>
</p>


```python
def fifo(ref, nf):
    frames = [-1] * nf
    front = faults = 0
    for page in ref:
        if page not in frames:
            frames[front] = page
            front = (front + 1) % nf
            faults += 1
    print(f"FIFO Page Faults  = {faults}")


def lru(ref, nf):
    frames = [-1] * nf
    last_used = [0] * nf
    filled = faults = 0
    for i, page in enumerate(ref):
        if page in frames:
            last_used[frames.index(page)] = i
        else:
            faults += 1
            if filled < nf:
                frames[filled] = page
                last_used[filled] = i
                filled += 1
            else:
                lru_idx = last_used.index(min(last_used))   # least recently used
                frames[lru_idx] = page
                last_used[lru_idx] = i
    print(f"LRU Page Faults   = {faults}")


def optimal(ref, nf):
    n = len(ref)
    frames = [-1] * nf
    filled = faults = 0
    for i, page in enumerate(ref):
        if page in frames[:filled]:
            continue
        faults += 1
        if filled < nf:
            frames[filled] = page
            filled += 1
            continue
        # replace the page used farthest in the future (or never used again)
        replace_idx, farthest = -1, -1
        for j in range(nf):
            k = i + 1
            while k < n and ref[k] != frames[j]:
                k += 1
            if k == n:                      # never used again
                replace_idx = j
                break
            if k > farthest:
                farthest, replace_idx = k, j
        frames[replace_idx] = page
    print(f"Optimal Page Faults = {faults}")


def main():
    n = int(input("Enter number of pages in reference string: "))
    ref = list(map(int, input("Enter the reference string: ").split()))[:n]
    nf = int(input("Enter number of frames: "))

    fifo(ref, nf)
    lru(ref, nf)
    optimal(ref, nf)


if __name__ == "__main__":
    main()
```
