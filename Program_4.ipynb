<p style="font-family: Times New Roman;

<b>AIM:</b> To implement the classical Producer-Consumer (Bounded Buffer) problem using POSIX semaphores and threads to achieve mutual exclusion and synchronization.
</p>



<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li> The Producer-Consumer problem involves two types of processes/threads sharing a fixed-size buffer: producers add items, and consumers remove them.</li>
<li>Synchronization must ensure the producer does not add to a full buffer and the consumer does not remove from an empty buffer, and that access to the shared buffer is mutually exclusive.</li>
<li>This is solved using three semaphores: 'empty' (counts empty slots, initialized to buffer size), 'full' (counts filled slots, initialized to 0), and 'mutex' (binary semaphore for mutual exclusion, initialized to 1).</li>
</ul>
</div>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>Program(PYTHON) and OUTPUT:</b>
</p>


```python
import threading
import time

BUFFER_SIZE = 5
ITEMS = 10

buffer = [0] * BUFFER_SIZE
in_idx = 0
out_idx = 0

empty = threading.Semaphore(BUFFER_SIZE)   # counts empty slots
full = threading.Semaphore(0)              # counts filled slots
mutex = threading.Semaphore(1)             # mutual exclusion


def producer():
    global in_idx
    for i in range(1, ITEMS + 1):
        empty.acquire()                    # sem_wait(&empty)
        mutex.acquire()                    # sem_wait(&mutex)
        buffer[in_idx] = i
        print(f"Producer produced item {i} at slot {in_idx}")
        in_idx = (in_idx + 1) % BUFFER_SIZE
        mutex.release()                    # sem_post(&mutex)
        full.release()                     # sem_post(&full)
        time.sleep(0.1)


def consumer():
    global out_idx
    for _ in range(ITEMS):
        full.acquire()
        mutex.acquire()
        item = buffer[out_idx]
        print(f"\t\tConsumer consumed item {item} from slot {out_idx}")
        out_idx = (out_idx + 1) % BUFFER_SIZE
        mutex.release()
        empty.release()
        time.sleep(0.15)


if __name__ == "__main__":
    prod = threading.Thread(target=producer)
    cons = threading.Thread(target=consumer)
    prod.start()
    cons.start()
    prod.join()
    cons.join()
    print("All items produced and consumed.")
```
