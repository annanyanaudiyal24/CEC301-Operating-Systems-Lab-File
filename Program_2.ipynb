<b>AIM:</b> To understand the concept of threads and write a multithreaded C program using the POSIX thread (pthread) library to create, synchronize and join threads.

<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li> A thread is the smallest unit of CPU execution within a process. Multiple threads within the same process share the same address space (code, data, heap and open files) but have their own stack, program counter and registers.</li>
<li>POSIX threads (pthreads) is a standard threading API for UNIX-like systems. Key functions include pthread_create() to create a new thread, pthread_join() to wait for a thread to finish, and pthread_exit() to terminate a thread.
</li>
<li>Multithreading improves responsiveness and allows parallel execution of independent tasks on multicore processors, but requires synchronization when threads share data.
</li>
</ul>
</div>


```python
import threading

NUM_THREADS = 4


def sum_array(data, idx):
    start, end = data[idx]["start"], data[idx]["end"]
    data[idx]["sum"] = sum(range(start, end))
    print(f"Thread {threading.get_ident() % 100000}: "
          f"sum[{start}..{end}) = {data[idx]['sum']}")


def main():
    n = 1_000_000
    chunk = n // NUM_THREADS
    data = []
    threads = []

    for i in range(NUM_THREADS):
        start = i * chunk
        end = n if i == NUM_THREADS - 1 else (i + 1) * chunk
        data.append({"start": start, "end": end, "sum": 0})

    for i in range(NUM_THREADS):
        t = threading.Thread(target=sum_array, args=(data, i))
        threads.append(t)
        t.start()                        # pthread_create

    total = 0
    for i in range(NUM_THREADS):
        threads[i].join()                # pthread_join
        total += data[i]["sum"]

    print(f"Total sum (0 to {n - 1}) computed using {NUM_THREADS} threads = {total}")


if __name__ == "__main__":
    main()
```
