<p style="font-family: Times New Roman;
font-size: 18px;">
<b>AIM:</b> To implement the Dining Philosophers problem using semaphores in a way that avoids deadlock, illustrating concurrency control among multiple processes sharing limited resources (forks).
</p>



<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li> Five philosophers sit around a table with five forks (one between each pair). Each philosopher alternately thinks and eats; to eat, a philosopher needs both the left and right fork.</li>
<li>If every philosopher picks up the left fork first, all may simultaneously hold one fork and wait forever for the right fork — a circular wait, causing deadlock.</li>
<li>Deadlock is avoided here by allowing at most (N-1) philosophers to try to pick up forks at the same time, using a counting semaphore, which breaks the circular-wait condition.</li>
</ul>
</div>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>Program(PYTHON) and OUTPUT:</b>
</p>


```python
import threading
import time

N = 5
forks = [threading.Semaphore(1) for _ in range(N)]
room = threading.Semaphore(N - 1)       # at most N-1 philosophers try to eat at once


def philosopher(pid):
    left = pid
    right = (pid + 1) % N
    for _ in range(2):
        print(f"Philosopher {pid} is thinking")
        time.sleep(0.1)

        room.acquire()                  # enter the "room" - avoids deadlock
        forks[left].acquire()           # pick up left fork
        forks[right].acquire()          # pick up right fork

        print(f"Philosopher {pid} is eating (fork {left} and fork {right})")
        time.sleep(0.15)

        forks[right].release()
        forks[left].release()
        room.release()
        print(f"Philosopher {pid} finished eating and put down forks\n")


if __name__ == "__main__":
    philos = [threading.Thread(target=philosopher, args=(i,)) for i in range(N)]
    for t in philos:
        t.start()
    for t in philos:
        t.join()
```
