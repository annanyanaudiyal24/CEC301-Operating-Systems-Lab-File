Contiguous Memory Allocation — <b>First Fit</b>, <b>Best Fit</b> and <b>Worst Fit</b>.
</p>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>AIM:</b> To implement First Fit, Best Fit and Worst Fit memory allocation strategies for contiguous memory partitioning and compare their block allocation and fragmentation.
</p>



<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li>In contiguous memory allocation, each process is allocated a single contiguous block of memory from a list of free memory blocks (holes).</li>
<li>First Fit allocates the first hole that is big enough — fast, but can leave many small unusable fragments at the start of memory.</li>
<li>Best Fit allocates the smallest hole that is big enough — minimizes leftover space per allocation but leaves many tiny unusable fragments over time and is slower (searches the entire list).</li>
<li>Worst Fit allocates the largest available hole — leaves the largest possible leftover fragment, which may remain usable for future requests.</li>
</ul>
</div>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>Program(PYTHON) and OUTPUT:</b>
</p>


```python
def print_table(title, psize, allocation):
    print(f"\n--- {title} ---")
    print("Process No.\tProcess Size\tBlock No.")
    for i, size in enumerate(psize):
        where = f"B{allocation[i] + 1}" if allocation[i] != -1 else "Not Allocated"
        print(f"P{i + 1}\t\t{size}\t\t{where}")


def first_fit(bsize, psize):
    block = bsize[:]
    allocation = [-1] * len(psize)
    for i, size in enumerate(psize):
        for j in range(len(block)):
            if block[j] >= size:
                allocation[i] = j
                block[j] -= size
                break
    print_table("First Fit", psize, allocation)


def best_fit(bsize, psize):
    block = bsize[:]
    allocation = [-1] * len(psize)
    for i, size in enumerate(psize):
        best = -1
        for j in range(len(block)):
            if block[j] >= size and (best == -1 or block[j] < block[best]):
                best = j
        if best != -1:
            allocation[i] = best
            block[best] -= size
    print_table("Best Fit", psize, allocation)


def worst_fit(bsize, psize):
    block = bsize[:]
    allocation = [-1] * len(psize)
    for i, size in enumerate(psize):
        worst = -1
        for j in range(len(block)):
            if block[j] >= size and (worst == -1 or block[j] > block[worst]):
                worst = j
        if worst != -1:
            allocation[i] = worst
            block[worst] -= size
    print_table("Worst Fit", psize, allocation)


def main():
    bn = int(input("Enter number of memory blocks: "))
    bsize = list(map(int, input("Enter sizes of blocks: ").split()))[:bn]
    pn = int(input("Enter number of processes: "))
    psize = list(map(int, input("Enter sizes of processes: ").split()))[:pn]

    first_fit(bsize, psize)
    best_fit(bsize, psize)
    worst_fit(bsize, psize)


if __name__ == "__main__":
    main()
```
