<p style="font-family: Times New Roman;

<b>AIM:</b> To implement the first Readers-Writers problem using semaphores, allowing multiple readers to read a shared resource concurrently while ensuring writers get exclusive access.
</p>



<div style="font-family: Times New Roman;
font-size: 18px;">
<p><b>Theory:</b></p>
<ul>
<li> The Readers-Writers problem models access to a shared resource (e.g., a file or database) by multiple reader and writer threads.</li>
<li>Rules: multiple readers may read simultaneously; a writer requires exclusive access (no other reader or writer may access the resource while it writes).</li>
<li>This classic (first) solution uses a semaphore 'wrt' for exclusive writer access, a semaphore 'mutex' to protect the shared counter 'read_count', which tracks the number of active readers. The resource is locked by the first reader and unlocked by the last reader; this variant may starve writers.</li>
</ul>
</div>



<p style="font-family: Times New Roman;
font-size: 18px;">
<b>Program(PYTHON) and OUTPUT:</b>
</p>


```python
import threading
import time

mutex = threading.Semaphore(1)    # protects read_count
wrt = threading.Semaphore(1)      # exclusive access for writers
read_count = 0
shared_data = 0


def reader(rid):
    global read_count
    mutex.acquire()
    read_count += 1
    if read_count == 1:
        wrt.acquire()             # first reader locks the resource
    mutex.release()

    print(f"Reader {rid}: read shared_data = {shared_data}")
    time.sleep(0.2)

    mutex.acquire()
    read_count -= 1
    if read_count == 0:
        wrt.release()             # last reader unlocks the resource
    mutex.release()


def writer(wid):
    global shared_data
    wrt.acquire()
    shared_data += 1
    print(f"Writer {wid}: wrote shared_data = {shared_data}")
    time.sleep(0.3)
    wrt.release()


if __name__ == "__main__":
    w = [None, None]
    r = []

    w[0] = threading.Thread(target=writer, args=(1,))
    w[0].start()
    for i in range(1, 6):
        t = threading.Thread(target=reader, args=(i,))
        r.append(t)
        t.start()
    w[1] = threading.Thread(target=writer, args=(2,))
    w[1].start()

    for t in r:
        t.join()
    w[0].join()
    w[1].join()
```
